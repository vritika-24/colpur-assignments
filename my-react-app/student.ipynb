```python
import sqlite3
conn=sqlite3.connect("studentt.db")
cursor=conn.cursor()
cursor.execute(""" ATTACH DATABASE 'marks.db' AS marksdb """)
cursor.execute(""" SELECT 
    s.enroll_no,
    s.st_name,
    s.st_branch,
    s.st_year,
    SUM(m.marks) AS total_marks,
    ROUND(AVG(m.marks), 2) AS percentage
FROM studentt s
JOIN  marksdb.marks m
ON s.enroll_no = m.enrol_no
GROUP BY 
    s.enroll_no,
    s.st_name,
    s.st_branch,
    s.st_year
ORDER BY s.enroll_no;""")
for row in cursor:
    print(row)
conn.close()
```
